## Setup

1. git clone the repository.

```bash
git clone https://github.com/u3ks/stateo2026
```

2. Install the required software environment with Conda. If you do not have Conda, install it by following these instructions (see [here](https://docs.conda.io/en/latest/miniconda.html)). Then create the environment ( this can take a few minutes ).
```bash
conda env create -n pangeo_earthcode -f pangeo_earthcode.yml
```

3. Launch a Jupyterlab notebook server from this environment.
```bash
conda activate pangeo_earthcode
jupyter lab
```

4. Open a web browser and connect to the Jupyterlab provided URL (you should see it in the jupyter lab command outputs), something like: http://localhost:8888/lab?token=42fac6733c6854578b981bca3abf5152.

## 2. Core aspects (in practice)


- 2.1 STAC + Data Model:

- 2.3 Examples:
    
    - SMARTCH4 - https://opensciencedata.esa.int/products/d13c-ch4-signatures-smart-ch4/collection
        - NetCDF grids to a single Zarr
        - Transformation + metadata: https://esa-earthcode.github.io/tutorials/smartch4-signatures-modified/
        - Data Access: https://github.com/ESA-EarthCODE/data_access_examples/tree/main/smart-ch4

	- S4AMAZONAS - https://opensciencedata.esa.int/products/sentinel1-amazonas-forest-loss/collection
        - Weekly tiffs tiles to a single zarr
        - Transformation + metadata: https://esa-earthcode.github.io/tutorials/sen4amazonas/
        - Data Access: https://github.com/ESA-EarthCODE/data_access_examples/tree/main/sen4amazonas
	
    - WAPOSAL - https://opensciencedata.esa.int/products/waposal-waves/collection
        - Monthly NetCDFs to multiple zarrs (trajectory data) 
        - Transformation + metadata: https://esa-earthcode.github.io/tutorials/waposal/
        - Data Access: https://github.com/ESA-EarthCODE/data_access_examples/tree/main/waposal

	- Lightning2EarthCare - https://opensciencedata.esa.int/products/earthcare-frame-lightning/collection
        - Thousands of NetCDFs to monthly parquet files (trajectory data) 
        - Transformation + metadata: https://github.com/piskalab/lightning2earthcare/tree/postprocessing/postprocess
        - Data Access: https://github.com/ESA-EarthCODE/data_access_examples/tree/main/lightning2earthcare
        - Dashboard: https://esa-earthcode.github.io/science-hub-dashboard/

    - Antarctica In-sync Cube - https://discourse-earthcode.eox.at/t/antartica-insync-data-cubes/107
        - Multiple datacubes
        - Data Access:  https://discourse-earthcode.eox.at/t/antartica-insync-data-cubes/107
        - Dashboard: https://esa-earthcode.github.io/polar-science-cluster-dashboard/explore/
